```python

url = 'https://hedgefollow.com/funds/Berkshire+Hathaway'

options = webdriver.FirefoxOptions()
# options.headless = True
options.add_argument('-headless')


driver = webdriver.Firefox(options=options)
driver.get(url)

time.sleep(3)

table = driver.find_element(By.ID, 'fund_holdings_equities')

headers = []
rows = []
#print('--- headers ---')
header_row = table.find_elements(By.TAG_NAME, 'tr')[0]
for cell in header_row.find_elements(By.TAG_NAME, 'th'):
    if(cell.get_attribute("style").find("display: none")<0):
      headers.append(cell.text)

# print(headers)

# print('--- data ---')
data_rows = table.find_elements(By.TAG_NAME, 'tr')[1:]
for row in data_rows:
    row_data = {}
    children = row.find_elements(By.TAG_NAME, 'td')
    if(len(children) > 1):
      for i, cell in enumerate(children):
          row_data[headers[i]] = cell.text
      rows.append(row_data)

driver.quit()

df_bh = pd.DataFrame(rows)


df_bh.head()
```

```python
# Eliminando las columnas innecesarias
df_bh = df_bh.drop(['Ownership Hist', 'Average Buy Price', 'Price History'], axis=1)
```
